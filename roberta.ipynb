```python

# 确保你已经安装了正确的版本
# pip install transformers==2.10.0
# pip install torch

from transformers import BertTokenizer, BertModel

# 或者使用 AutoTokenizer, AutoModel，更具通用性
# from transformers import AutoTokenizer, AutoModel

# 模型标识符
model_name = 'hfl/chinese-roberta-wwm-ext'

print(f"正在从 '{model_name}' 加载 Tokenizer...")
# 加载 Tokenizer
# 对于这个模型，使用 BertTokenizer 是最直接的
tokenizer = BertTokenizer.from_pretrained(model_name)

print(f"正在从 '{model_name}' 加载预训练模型...")
# 加载预训练的基础模型（不含任何特定任务的头部）
model = BertModel.from_pretrained(model_name)

# -------------------
# 一个简单的使用示例
# -------------------
text = "你好，世界！"

# 使用 tokenizer 对文本进行编码
# 在 2.10.0 版本中，tokenizer() 返回的是一个字典
# inputs = tokenizer(text, return_tensors='pt') # pt 代表 PyTorch tensors
inputs = tokenizer.encode_plus(text, return_tensors='pt')

print("\n编码后的输入:", inputs)

# 将编码后的输入传入模型
# 模型会返回一个元组，第一个元素是 last_hidden_state
outputs = model(**inputs)
last_hidden_state = outputs[0]

print("\n模型输出的 last_hidden_state 维度:", last_hidden_state.shape)
# 输出维度通常是 [batch_size, sequence_length, hidden_size]
# 对于 "你好，世界！"，分词后可能是 ['[CLS]', '你', '好', '，', '世', '界', '！', '[SEP]'] -> sequence_length=8
# RoBERTa-wwm-ext 的 hidden_size 是 768
```
